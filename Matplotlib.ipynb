Setup

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Carga de datos

Graficación de la recta

```python
X = np.arange(0,15,0.5)
y = -2 * X + 6
print(X)
print(y)

plt.plot(X, y, "go--")
plt.show()
```

```python
X = [0, 0, 1, 1, 0]
y = [0, 1, 1, 0, 0]
print(X)
print(y)

plt.plot(X, y, "ro--")
plt.title("Cubo")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
X = np.linspace(0.1, 7, 50)
y = np.cos(X) / X
print(X)
print(y)
plt.plot(X, y, "g")
plt.title("Cos(X)/X")
plt.xlabel("Ekis")
plt.ylabel("Ye")
plt.show()
#
```

```python
n = 120
X = np.linspace(0, 100, n)
y = 2.0 * X + np.random.normal(0, 25, n)
y_real = 2.0 * X
fig = plt.figure(figsize=(6,4))
ax = fig.gca()
ax.scatter(X, y , alpha=0.75)
ax.plot(X, y_real, "r--")
ax.set_title("dispersión: relación entre 2 variables")
ax.set_xlabel("X")
ax.set_ylabel("y")
ax.grid(True, linestyle="--", alpha = 0.4)
plt.show()
```

Generar una función que genere una nube negativa

```python
#Generar una función que genere una nube negativa
n = 120
X = np.linspace(0, 100, n)
y = 2.0 * -X + np.random.normal(0, 25, n)
y_real = 2.0 * -X
fig = plt.figure(figsize=(6,4))
ax = fig.gca()
ax.scatter(X, y , alpha=0.75)
ax.plot(X, y_real, "r--")
ax.set_title("dispersión: relación entre 2 variables")
ax.set_xlabel("X")
ax.set_ylabel("y")
ax.grid(True, linestyle="--", alpha = 0.4)
plt.show()
```

```python
# histograma
muestra = np.random.normal(0, 1, 10000)
fig = plt.figure(figsize=(6,4))
ax = plt.gca()
ax.hist(muestra, bins = 20, alpha = 0.9)
ax.set_title("Distribución de valores")
ax.set_xlabel("Valor")
ax.set_ylabel("Frecuencia")

ax.grid(True, linestyle="--", axis = 'y', alpha = 0.4)
plt.show()

#Calcular el residual
X = muestra
y = 2.0 * X + np.random.normal(0, 1, 10000)
y_real = 2.0 * X

residual = y - y_real

print("\nResiduo:\n")
residual_df = pd.DataFrame(residual, columns=['Residual'])

display(residual_df)
```
