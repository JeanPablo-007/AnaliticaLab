```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Set 1.
#Paso 1. Datos
cats = ['Norte', 'Sur', 'Este', 'Oeste']
conteos = np.array([120, 90, 150, 110])

#Paso 2. Figura y eje
fig = plt.figure(figsize=(6,4))
ax = plt.gca()

#Paso 3. Gráfico de barras
bars = ax.bar(cats, conteos)

# Añadir los números encima de las barras
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, yval + 1, round(yval), ha='center', va='bottom') # +1 para un pequeño margen

#Paso 4. Etiqueta y Grid
ax.set_title("Distribución de categorías")
ax.set_xlabel("Categoría")
ax.set_ylabel("Valor")
ax.grid(True, linestyle="--", alpha = 0.4)

plt.show()
```

```python
# Set 2.
#Paso 1. Datos
n = 120
X = np.linspace(0, 100, n)
y = 2.0 * X + np.random.normal(0, 25, n)  # pendiente ~2 + ruido
y_real = 2.0 * X

#Paso 2. Figura y eje
fig = plt.figure(figsize=(6,4))
ax = fig.gca()

#Paso 3. Selección de gráfico
ax.scatter(X, y , alpha=0.75)

#Paso 4. Etiqueta y Grid
ax.plot(X, y_real, "r--")
ax.set_title("dispersión: relación entre 2 variables")
ax.set_xlabel("X")
ax.set_ylabel("y")
ax.grid(True, linestyle="--", alpha = 0.4)

#Paso 5. Mostrar gráfico
plt.show()
```

```python
# Set 3.
#Paso 1. Datos
muestra = np.random.normal(loc=50, scale=10, size=1000) + np.random.exponential(scale=2, size=1000)*0.5

#Paso 2. Figura y eje
fig = plt.figure(figsize=(6,4))
ax = plt.gca()

#Paso 3. Selección de gráfico
ax.hist(muestra, bins = 20, alpha = 0.9)

#Paso 4. Etiqueta y Grid
ax.set_title("Distribución de valores")
ax.set_xlabel("Valor")
ax.set_ylabel("Frecuencia")

ax.grid(True, linestyle="--", axis = 'y', alpha = 0.4)
plt.show()

# Paso 5. Calcular el residual
X = muestra
y = 2.0 * X + np.random.normal(0, 1, 1000)
y_real = 2.0 * X

residual = y - y_real

print("\nResiduo:\n")
residual_df = pd.DataFrame(residual, columns=['Residual'])

#Paso 6. Mostrar gráfico
display(residual_df)
```

```python
# Set 4.
#Paso 1. Datos
t = np.arange(1, 25)
tendencia = 5 + 1.2 * t                       # componente de tendencia (creciente)
estacional = 4 * np.sin(2 * np.pi * t / 12)  # componente estacional anual (periodo=12)
ruido = np.random.normal(0, 2, size=t.shape) # componente aleatoria (ruido)
serie = tendencia + estacional + ruido

#Paso 2. Figura y eje
fig = plt.figure(figsize=(6,4))
ax = plt.gca()

#Paso 3. Selección de gráfico
ax.plot(t, serie, 'go-')

#Paso 4. Etiqueta y Grid
ax.set_title("Serie de tiempo")
ax.set_xlabel("Tiempo")
ax.set_ylabel("Valor")
ax.grid(True, linestyle="--", alpha = 0.4)

#Paso 5. Mostrar gráfico
plt.show()
```

```python
# Set 5.
#Paso 1. Datos
df = pd.DataFrame({
    'ingresos': np.random.normal(50, 10, 200),
    'gasto_marketing': np.random.normal(25, 5, 200),
    'clientes_nuevos': np.random.poisson(12, 200) + np.random.normal(0, 1, 200),
    'satisfaccion': np.clip(np.random.normal(3.5, 0.7, 200), 1, 5)
})
# Inducimos una variable dependiente combinada (ventas).
#Paso 2.
df['ventas'] = 2.5 * df['ingresos'] + 1.1 * df['gasto_marketing'] + np.random.normal(0, 8, 200)

#Paso 3) Correlación
corr = df.corr(numeric_only=True)

#Paso 4) Gráfico
fig = plt.figure(figsize=(6,5))
ax = plt.gca()

#Paso 5) Mapa de Calor
im = ax.imshow(corr, cmap='coolwarm')
ax.set_title("Mapa de Calor de Correlación entre variables")
ax.set_xticks(np.arange(len(corr.columns)))
ax.set_yticks(np.arange(len(corr.columns)))
ax.set_xticklabels(corr.columns, rotation = 45, ha = 'right')
ax.set_yticklabels(corr.columns)
plt.colorbar(im, ax = ax, fraction = 0.046, pad = 0.04)
plt.tight_layout()

#Paso 6) Mostrar
plt.show()
```

```python
# Set 6.
#Paso 1. Datos
categorias = ['A', 'B', 'C', 'D']
valores = np.array([35, 25, 22, 18])

#Paso 2. Figura y eje
fig = plt.figure(figsize=(6,6))
ax = plt.gca()

#Paso 3. Seleccionar Gráfico
ax.pie(valores, labels = categorias, autopct = "%1.1f%%")

#Paso 4. Etiquetas
ax.axis("equal")
ax.set_title("Distribución de categorías")
ax.set_xlabel("Categoría")
ax.set_ylabel("Valor")

#Paso 5. Mostrar gráfico
plt.show()
```

```python
# Set 8
#Paso 1. Datos
n = 60
X = np.random.uniform(10, 100, size=n)
Y = 1.5 * X + np.random.normal(0, 15, size=n)
S = np.random.uniform(50, 1500, size=n)

#Paso 2)
fig = plt.figure(figsize=(6,4))
ax = plt.gca()

#Paso 3. Definir gráfico
ax.scatter(X, Y, s = S, alpha = 0.5)

#Paso 4. Metadatos y rejilla
ax.set_title("Gráfico de Burbujas: relación entre 3 variables")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.grid(True, linestyle="--", alpha = 0.4)

#Paso 5. Mostrar
plt.show()
```
