```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
#Gráfica de barras de variables ordinales
cats = ['1ro', '2do', '3ro', '4to']
conteos = np.array([20, 10, 15, 30])

#Paso 2. Figura y eje
fig = plt.figure(figsize=(7,5))
ax = plt.gca()

#Paso 3. Gráfico de barras
bars = ax.bar(cats, conteos)

# Añadir los números encima de las barras
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, yval, round(yval), ha='center', va='bottom')

#Paso 4. Etiqueta y Grid
ax.set_title("Número de alumnos inscritos")
ax.set_xlabel("Grado")
ax.set_ylabel("Cantidad")
ax.grid(True, linestyle="--", alpha = 0.4)

plt.show()
```

```python
#Gráfica de barras de variables nominales
cats = ['competitivos', 'casuales', 'acción', 'terror']
conteos = np.array([8, 7, 5, 3])


#Paso 2. Figura y eje
fig = plt.figure(figsize=(7,5))
ax = plt.gca()

#Paso 3. Gráfico de barras
bars = ax.bar(cats, conteos)

# Añadir los números encima de las barras
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, yval, round(yval), ha='center', va='bottom')

#Paso 4. Etiqueta y Grid
ax.set_title("Preferencias de videojuegos de las personas")
ax.set_xlabel("Categoría")
ax.set_ylabel("Valor")
ax.grid(True, linestyle="--", alpha = 0.4)

plt.show()
```
